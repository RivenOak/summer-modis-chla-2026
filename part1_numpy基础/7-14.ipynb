# NumPy速查代码卡片

### 1.数组创建

```python
import numpy as np

# 从列表创建
a = np.array([[1, 2], [3, 4]])
print("从列表创建:\n", a)

# 全零 / 全一
zeros = np.zeros((2, 3))
ones  = np.ones((3, 2))
print("\n全零:\n", zeros)
print("全一:\n", ones)

# 等差序列
arange_arr = np.arange(0, 12, 3)      # [0, 3, 6, 9]
linspace_arr = np.linspace(0, 1, 5)   # [0., 0.25, 0.5, 0.75, 1.]
print("\narange:", arange_arr)
print("linspace:", linspace_arr)

# 随机数组
rand_arr = np.random.rand(2, 4)
print("\n随机数组:\n", rand_arr)
```

### 2.数组属性与类型

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])

print("形状:", a.shape)       # (2, 3)
print("维度数:", a.ndim)      # 2
print("数据类型:", a.dtype)   # int64（默认）

# 改变数据类型
b = a.astype(np.float32)
print("转换后类型:", b.dtype)  # float32
```

### 向量化运算与广播

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

# 元素级运算
print("加法:", a + b)         # [11 22 33]
print("乘法:", a * b)         # [10 40 90]
print("幂运算:", a ** 2)      # [1 4 9]

# 广播：数组与标量
print("广播加法:", a + 5)     # [6 7 8]

# 矩阵乘法（dot product）
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print("\n矩阵乘法:\n", A @ B)  # 等价于 np.dot(A, B)
```

### 5.统计聚合（含axis用法）

```python
data = np.array([[1, 2, 3],
                 [4, 5, 6]])

print("全局求和:", data.sum())        # 21
print("全局均值:", data.mean())       # 3.5
print("全局最大值:", data.max())      # 6

# axis 参数：0=沿列方向（跨行），1=沿行方向（跨列）
print("每列求和:", data.sum(axis=0))  # [5 7 9]
print("每行均值:", data.mean(axis=1)) # [2. 5.]
```
