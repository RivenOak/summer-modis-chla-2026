```python
import matplotlib.pyplot as plt
import numpy as np

#设置字体为黑体（Windows常用）或微软雅黑，以便支持中文显示
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei']
#解决保存图像时负号'-'显示为方块的问题
plt.rcParams['axes.unicode_minus'] = False

x1 =np.array([2023,2024,2025,2026])
y1 = np.array([12,27,68,35])

x2 = np.array([2019,2020,2021,2022])
y2 =np.array([24,45,76,34])

plt.title('表格一',fontsize=20,
                   fontweight='bold',
                   color='#2d4cfc')


plt.xlabel('年份',fontsize=20,
                  fontweight='bold',
                  color='#2d4cfc')

plt.ylabel('数量',fontsize=20,
                  fontweight='bold',
                  color='#2d4cfc')

line_style= dict(marker = '.',
                 markersize=20,
                 markerfacecolor='yellow',
                 markeredgecolor='yellow',
                 linestyle='dashed',#默认是实线solid，虚线是dashed,点线是dotted,虚点线是dashdot,或者'None'是无线  
                 linewidth=2,
                 color='#1c5bfc')
#设置刻度参数
plt.tick_params(axis='both',
                colors='#2d4cfc')

plt.plot(x1,y1,**line_style) 

plt.plot(x1,y2)

plt.plot(x2,y1,**line_style)
#x轴仅显示数值的刻度
all_ticks = np.concatenate([x2, x1]) 
plt.xticks(all_ticks)
#添加网格线
plt.grid(axis="both",
         linewidth=1.5,
         color='lightgray',
         linestyle='dashed')
plt.show()
```

柱状图bar chart

```python
#创建一个python列表
categories=['谷物Grains','水果Fruit','蔬菜Vegetables','蛋白质Protein','乳制品Dairy','甜品Sweets']
values=np.array([3,2,3,5,4,1])

plt.bar(categories,values,color='green')
#水平柱形图：plt.barh(categories,values,color='green')

plt.title('每日摄入量')

plt.xlabel('食物')

plt.ylabel('比例')

#旋转45度，并设置为右对齐（ha='right'），这样标签根部会对齐刻度线，看起来更整齐
plt.xticks(rotation=45, ha='right')

plt.show()
```

饼状图pie

```python
colors = ['#b6de71','#dece71','#71deb8','#71d8de','#de71b2','#a271de']

plt.pie(values,labels=categories,
               autopct='%1.1f%%',
               colors=colors,
               explode=[0,0.1,0,0.1,0,0],
               shadow=True,
               startangle=90)#旋转多少角度

plt.show()
```
