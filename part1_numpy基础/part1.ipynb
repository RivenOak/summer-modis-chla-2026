## 1.数组创建与基础操作
### 1.1创建不同形状的数组

```python
zeros_arr = np.zeros((3, 4))          # 全零数组
ones_arr = np.ones((2, 5))            # 全一数组
rand_arr = np.random.rand(3, 3)       # 均匀分布 [0,1) 随机数组
eye_mat = np.eye(4)                   # 单位矩阵

print("zeros (3x4):\n", zeros_arr)
print("\nones (2x5):\n", ones_arr)
print("\nrandom (3x3):\n", rand_arr)
print("\neye (4x4):\n", eye_mat)
```

### 1.2查看属性

```python
arr = np.array([[1, 2, 3], 
                [4, 5, 6]])
print(f"形状: {arr.shape}")
print(f"数据类型: {arr.dtype}")
print(f"元素个数: {arr.size}")
print(f"维度: {arr.ndim}")
```

### 1.3重塑与索引

```python
arr_flat = arr.flatten()
print("展平:", arr_flat)

arr_reshaped = arr.reshape(3, 2)
print("重塑为 3x2:\n", arr_reshaped)

# 索引
print("第一行第二列:", arr[0, 1])   # 输出 2
print("第二行所有列:", arr[1, :])   # 输出 [4 5 6]
```

## 2.矩阵运算  
### 2.1 矩阵乘法

```python
A = np.array([[2, 1],
              [1, 3]])
B = np.array([[1, 2],
              [3, 4]])

print("A @ B:\n", A @ B)          # 等价于 np.dot(A, B)
print("B @ A:\n", np.dot(B,A))          # 矩阵乘法不满足交换律
```

### 2.2行列式与逆矩阵

```python
det_A = np.linalg.det(A)
inv_A = np.linalg.inv(A)

print(f"A 的行列式: {det_A:.4f}")
print("A 的逆矩阵:\n", inv_A)

# 验证 A * A_inv = I
identity_check = A @ inv_A
print("\nA * A_inv:\n", identity_check)
print("是否接近单位矩阵:", np.allclose(identity_check, np.eye(2))) #如果两个数组在容差范围内逐元素相等，则返回 True
```

### 2.3解线性方程组 Ax = b

```python
b = np.array([5, 6])
x = np.linalg.solve(A, b)
print("方程的解 x =", x)

# 验证 A @ x == b
print("验证 A @ x =", A @ x)
```

## 3、特征值与特征向量  
### 3.1计算特征值和特征向量

特征值表示伸缩，特征向量表示不变的方向

```python
eigenvalues, eigenvectors = np.linalg.eig(A)
print("特征值:", eigenvalues)
print("特征向量:\n", eigenvectors)

# 验证 Av = λv
for i in range(len(eigenvalues)):
    v = eigenvectors[:, i]
    lam = eigenvalues[i]
    lhs = A @ v
    rhs = lam * v
    print(f"特征值 {lam:.2f}: A*v ≈ λ*v ? {np.allclose(lhs, rhs)}")
```

### 3.2对称矩阵的特征分解（实对称矩阵特征值均为实数）

```python
sym_mat = np.array([[3, 1, 0],
                    [1, 4, 2],
                    [0, 2, 5]])
eigvals_sym, eigvecs_sym = np.linalg.eigh(sym_mat)  # eigh 专门用于对称/厄米矩阵
print("对称矩阵特征值:", eigvals_sym)
print("特征向量正交性检查:\n", eigvecs_sym.T @ eigvecs_sym)  # 应为单位矩阵
```
